# **Analisis Statistik **##

```python
from scipy import stats
from statsmodels.stats.diagnostic import lilliefors
```

```python
kolumn_numerik = df.select_dtypes(include=[np.number]).columns

alpha = 0.05
hasil_normalitas = []

for col in kolumn_numerik:
    p_lilliefors = np.nan  # Initialize p-value for current column
    try:
        # The lilliefors test function returns statistic and p-value
        _, p_lilliefors = lilliefors(df[col])
    except Exception as e:
        print(f"Error calculating Lilliefors test for column '{col}': {e}")
        # p_lilliefors remains np.nan

    hasil_normalitas.append({
        "nama kolom": col,
        "jumlah sampel (n)": len(df[col]),
        "nilai minimum": df[col].min(),
        "nilai maksimum": df[col].max(),
        "p-value (Lilliefors)": round(p_lilliefors, 5)
        if pd.notna(p_lilliefors)
        else "-",
    })

df_hasil_normalitas = pd.DataFrame(hasil_normalitas)
display(df_hasil_normalitas)
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. Klasifikasikan Kolom Berdasarkan Tipe Datanya
# Variabel Kontinu/Numerik (Diuji dengan Mann-Whitney U)
kolom_kontinu = [
    'Age',
    'Income',
    'Travel_Frequency',
    'Vacation_Budget',
    'Proximity_to_Mountains',
    'Proximity_to_Beaches',
]

# Variabel Biner/Kategorikal (Diuji dengan Chi-Square)
kolom_biner = ['Pets', 'Environmental_Concerns']

target = 'Preference'  # 1 = Pantai, 0 = Non-Pantai
alpha = 0.05
hasil_pengujian = []

# ==========================================
# A. UJI MANN-WHITNEY U (Untuk Variabel Numerik)
# ==========================================
for col in kolom_kontinu:
    kelompok_0 = df[df[target] == 0][col].dropna()  # Non-Pantai
    kelompok_1 = df[df[target] == 1][col].dropna()  # Pantai

    # Jalankan Uji Mann-Whitney U
    stat_mw, p_val_mw = stats.mannwhitneyu(
        kelompok_1, kelompok_0, alternative='two-sided'
    )

    status = (
        'Signifikan (Ada Beda)' if p_val_mw < alpha else 'Tidak Signifikan'
    )

    hasil_pengujian.append({
        'Nama Variabel': col,
        'Jenis Uji': 'Mann-Whitney U',
        'Statistik Uji': round(stat_mw, 2),
        'p-value': round(p_val_mw, 5),
        'Kesimpulan': status,
    })

# ==========================================
# B. UJI CHI-SQUARE (Untuk Variabel Biner)
# ==========================================
for col in kolom_biner:
    # Buat Tabel Silang (Crosstab)
    crosstab = pd.crosstab(df[col], df[target])

    # Jalankan Uji Chi-Square
    chi2_stat, p_val_chi2, dof, _ = stats.chi2_contingency(crosstab)

    status = (
        'Signifikan (Ada Hubungan)'
        if p_val_chi2 < alpha
        else 'Tidak Signifikan'
    )

    hasil_pengujian.append({
        'Nama Variabel': col,
        'Jenis Uji': 'Chi-Square',
        'Statistik Uji': round(chi2_stat, 2),
        'p-value': round(p_val_chi2, 5),
        'Kesimpulan': status,
    })

# ==========================================
# C. TAMPILKAN HASIL DALAM TABEL RAPI
# ==========================================
df_hasil_nonparametrik = pd.DataFrame(hasil_pengujian)
display(df_hasil_nonparametrik)
```

```python
from scipy.stats import ttest_ind

# Separate the 'Proximity_to_Beaches' data based on 'Preference'
proximity_beach_preference_0 = df[df['Preference'] == 0]['Proximity_to_Beaches']
proximity_beach_preference_1 = df[df['Preference'] == 1]['Proximity_to_Beaches']

# Perform independent samples t-test
ttest_result = ttest_ind(proximity_beach_preference_0, proximity_beach_preference_1)

print("Independent samples T-test result:")
print(f"T-statistic: {ttest_result.statistic:.3f}")
print(f"P-value: {ttest_result.pvalue:.3f}")

# Interpret the result based on the p-value and a significance level (e.g., 0.05)
alpha = 0.05
if ttest_result.pvalue < alpha:
    print("Tolak H0 : Jarak atau kedekatan wisata berpengaruh signifikan")
else:
    print("Terima H0 : Jarak atau kedekatan wisata tidak berpengaruh signifikan")
```

```python
import xgboost as xgb
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn import model_selection
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score,fbeta_score
from sklearn.metrics import roc_curve, auc, roc_auc_score
from sklearn.model_selection import GridSearchCV
```

```python
models = []
models.append (('Logistics Regression',LogisticRegression(solver='liblinear', random_state=0,
                                                          class_weight='balanced')))
models.append (('Xgboost Classfier',xgb.XGBClassifier(objective='binary:logistic', eval_metric='logloss', random_state =0)))
models.append (('KNN', KNeighborsClassifier()))
models.append (('SVM', SVC(kernel='linear')))
models.append (('Naive Bayes', GaussianNB()))
```

```python
col = ['Algorithm','Mean Accuracy', 'Mean AUC', 'STD Accuracy', 'STD AUC']
results = pd.DataFrame(columns=col)

kfold = KFold(n_splits=10, shuffle=True, random_state=0)

for name, model in models:
    cv_results = cross_val_score(model, x_train, y_train, cv=kfold, scoring='accuracy')
    cv_auc = cross_val_score(model, x_train, y_train, cv=kfold, scoring='roc_auc')

    results.loc[name] = [
        name,
        round(cv_results.mean() * 100, 4),
        round(cv_auc.mean()* 100, 4),
        round(cv_results.std()* 1000,4),
        round(cv_auc.std()*1000,4),
    ]

results = results.sort_values(by='Mean AUC', ascending=False)
results
```

**Ekstrak Model**

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier

#Definisikan Kolom
categorical_ordinal = ['Education_Level']
# Corrected the categories for OrdinalEncoder
Education_cat_order = [['high school', 'bachelor', 'master', 'doctorate']]

categorical_nominal = [col for col in df.columns
                       if df[col].dtype == 'object' and col not in categorical_ordinal]
numerical_col = [col for col in df.columns if col not in categorical_ordinal + categorical_nominal + ['Preference']] # Corrected 'preference' to 'Preference'

#Definisikan Transformer
ordinal_transformer = OrdinalEncoder(categories=Education_cat_order)
onehit_transformer = OneHotEncoder(handle_unknown='ignore')
scaler_transformer = StandardScaler()

preprocessor = ColumnTransformer(
    transformers=[
        ('ordinal', ordinal_transformer, categorical_ordinal),
        ('onehot', onehit_transformer, categorical_nominal),
        ('scaler', scaler_transformer, numerical_col)
    ],
    remainder='passthrough' # Changed remainder to 'passthrough' to keep other columns
)

# Define the model
model = LogisticRegression(random_state=0)

# Create the pipeline
pipeline = ImbPipeline(steps=[('preprocessor', preprocessor),
                              ('smote', SMOTE(random_state=42)), # Added SMOTE to the pipeline
                              ('classifier', model)])


# Train-test split
x = df.drop('Preference', axis=1)
y = df['Preference']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

# Fit Model
pipeline.fit(x_train, y_train)

#evaluasi Model
y_pred = pipeline.predict(x_test)
```
