```python
from scipy.stats import shapiro

df_numeric = df_1.select_dtypes(include=['number'])

hasil_uji = []

for kolom in df_numeric.columns:
    data_bersih = df_numeric[kolom].dropna()
    stat, p_value = shapiro(data_bersih)
    
    status = 'Normal' if p_value > 0.05 else 'Tidak Normal'
    
    hasil_uji.append({
        'Kolom': kolom,
        'Statistic': stat,
        'p-value': p_value,
        'Kesimpulan': status
    })

df_hasil = pd.DataFrame(hasil_uji)

df_hasil

```

```python
import pandas as pd
from scipy import stats

kolom_grup = 'Variabel'  

kolom_numerik = df.select_dtypes(include=['number']).columns

hasil_kruskal = []

for kolom in kolom_numerik:
    
    groups = [group[kolom].dropna() for name, group in df.groupby(kolom_grup)]
    
    if len(groups) > 1:
        stat, p_value = stats.kruskal(*groups)
        
        status = 'Signifikan (Ada Perbedaan)' if p_value < 0.05 else 'Tidak Signifikan (Tidak Ada Perbedaan)'
        
        hasil_kruskal.append({
            'Kolom Numerik': kolom,
            'Statistic': stat,
            'p-value': p_value,
            'Kesimpulan': status
        })

df_kruskal = pd.DataFrame(hasil_kruskal)
df_kruskal
```

**Modelling Data**

```python
target = df_1[['Category Visual Attention ']]
df_tmp = df_1.drop(columns=['Category Visual Attention ',])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df_tmp, target, test_size=0.2, random_state=42)
```

```python
print("Training set size:", x_train.shape)
print("Test set size:", x_test.shape)
print("Training set size:", x_train.shape)
print("Test set size:", x_test.shape)
```

```python
print(x_train.dtypes)
```

```python
# Menghapus kolom 'Responden' dari x_train dan x_test
x_train = x_train.drop(columns=['Responden'])
x_test = x_test.drop(columns=['Responden'])
```

```python
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler

from sklearn.naive_bayes import GaussianNB
from sklearn import model_selection
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import mean_squared_error, r2_score
```

```python
method = StandardScaler()
x_train_scaled = pd.DataFrame(method.fit_transform(x_train), columns=x_train.columns)
x_train = x_train_scaled

xtest_scaled = pd.DataFrame(method.transform(x_test), columns=x_test.columns)
x_test = xtest_scaled
```

```python
models = []
models.append (('Xgboost Classfier',xgb.XGBClassifier(objective='binary:logistic', eval_metric='logloss', random_state =0)))

models.append(('Naive Bayes',
               GaussianNB()))
print(models,"Berhasil Membuat List Model")
```

```python
import xgboost as xgb
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn import model_selection
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score,fbeta_score
from sklearn.metrics import roc_curve, auc, roc_auc_score
from sklearn.model_selection import GridSearchCV
```

```python

import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_validate

col = [
    'Algorithm',
    'Mean Accuracy',
    'Mean AUC',
    'STD Accuracy',
    'STD AUC'
]

results = pd.DataFrame(columns=col)

kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=0
)

for name, model in models:
    cv_results = cross_validate(
        model,
        x_train,
        y_train,
        cv=kfold,
        scoring={
            'accuracy': 'accuracy',
            'auc': 'roc_auc_ovr'
        },
        error_score='raise'
    )

    results.loc[len(results)] = [
        name,
        round(cv_results['test_accuracy'].mean() * 100, 4),
        round(cv_results['test_auc'].mean() * 100, 4),
        round(cv_results['test_accuracy'].std() * 100, 4),
        round(cv_results['test_auc'].std() * 100, 4)
    ]

results = results.sort_values(
    by='Mean AUC',
    ascending=False
).reset_index(drop=True)

results
```

```python
classfier = xgb.XGBClassifier(random_state=0)
classfier.fit(x_train, y_train)
y_pred = classfier.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix (y_test, y_pred)
cm_display = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classfier.classes_)
cm_display.plot(colorbar=False)
ax = cm_display.ax_
ax.tick_params(axis='both', labelsize=15)

for text in ax.texts:
    text.set_fontsize(15)
    text.set_fontweight('bold')
plt.show()
```

```python
from sklearn.metrics import classification_report, confusion_matrix

# y_true adalah label asli, y_pred adalah hasil prediksi model Anda
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Prediksi dengan model
y_train_pred = classfier.predict(x_train)
y_test_pred  = classfier.predict(x_test)

# Hitung akurasi
train_acc = accuracy_score(y_train, y_train_pred)
test_acc  = accuracy_score(y_test, y_test_pred)

print("Akurasi Training :", train_acc)
print("Akurasi Testing  :", test_acc)

# Laporan detail
print("\nClassification Report (Test):")
print(classification_report(y_test, y_test_pred))
```
